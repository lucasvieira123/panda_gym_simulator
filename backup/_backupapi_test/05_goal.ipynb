# 05 — Goal
Controle de goals em tempo de execução via `PUT /goal`.

## Pré-requisito
```bash
cd managing && python main.py
```

## Ações disponíveis
| `action`        | Descrição                          | Campos obrigatórios  |
|-----------------|------------------------------------|----------------------|
| `move_target`   | Move uma esfera de goal            | `name`, `position`   |
| `set_goal_mode` | Troca o modo de seleção de goal    | `mode`               |

## Nomes das esferas
| Nome       | Descrição                            |
|------------|--------------------------------------|
| `target`   | Primeiro goal (ou único)             |
| `target_1` | Segundo goal                         |
| `target_2` | Terceiro goal                        |
| `target_N` | N+1-ésimo goal                       |

````python
import requests
import json
import time

BASE_URL = "http://localhost:8000"

def get(endpoint: str) -> dict:
    r = requests.get(f"{BASE_URL}{endpoint}", timeout=3)
    r.raise_for_status()
    return r.json()

def put(endpoint: str, body: dict) -> dict:
    r = requests.put(f"{BASE_URL}{endpoint}", json=body, timeout=3)
    r.raise_for_status()
    return r.json()

def pp(data: dict):
    print(json.dumps(data, indent=2, default=str))

print("Setup OK")
````

---
## `move_target` — mover um goal individual

````python
# Mover o primeiro goal para uma nova posição
response = put("/goal", {
    "action":   "move_target",
    "name":     "target",
    "position": [0.20, 0.0, 0.02]
})
print(response)
````

````python
# Reposicionar todos os goals da sequência
put("/goal", {"action": "move_target", "name": "target",   "position": [ 0.15,  0.0,  0.02]})
put("/goal", {"action": "move_target", "name": "target_1", "position": [ 0.15,  0.10, 0.02]})
put("/goal", {"action": "move_target", "name": "target_2", "position": [-0.10,  0.0,  0.02]})
put("/goal", {"action": "move_target", "name": "target_3", "position": [-0.10,  0.10, 0.02]})
print("Goals reposicionados.")
````

---
## `set_goal_mode` — trocar o modo de seleção de goal

| `mode`          | Comportamento                                               |
|-----------------|-------------------------------------------------------------|
| `goal_options`  | Vai sempre para o target **mais próximo** da posição atual  |
| `goal_sequence` | Segue os targets **em ordem**, um por vez                   |
| `goal_set`      | Alcança **todos** os targets, escolhendo o mais próximo     |

> O reset interno (índice e completados) acontece automaticamente ao trocar o modo.  
> O modo persiste mesmo ao trocar de task via `PUT /task`.

````python
# goal_options — sempre o target mais próximo
response = put("/goal", {
    "action": "set_goal_mode",
    "mode":   "goal_options"
})
print(response)
````

````python
# goal_sequence — seguir em ordem
response = put("/goal", {
    "action": "set_goal_mode",
    "mode":   "goal_sequence"
})
print(response)
````

````python
# goal_set — alcançar todos, qualquer ordem
response = put("/goal", {
    "action": "set_goal_mode",
    "mode":   "goal_set"
})
print(response)
````
